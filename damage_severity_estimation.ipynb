##Detecting Label Format

```python
label_dir = f'{PROJECT_ROOT}/labels/train'
label_files = [f for f in os.listdir(label_dir) if f.endswith('.txt')]
sample_files = random.sample(label_files, min(20, len(label_files)))

line_lengths = []
for fname in sample_files:
    with open(os.path.join(label_dir, fname)) as f:
        for line in f:
            parts = line.strip().split()
            if parts:
                line_lengths.append(len(parts))

if not line_lengths:
    raise ValueError('No label content found — check that label .txt files are not empty')

max_len = max(line_lengths)
is_segmentation = max_len > 5

print(f'Sampled {len(sample_files)} label files, {len(line_lengths)} annotation lines')
print(f'Max fields per line: {max_len}')
print(f'Detected format: {"SEGMENTATION (polygon masks)" if is_segmentation else "DETECTION (bounding boxes)"}')

MODEL_NAME = 'yolov8n-seg.pt' if is_segmentation else 'yolov8n.pt'
print(f'\nModel to use: {MODEL_NAME}')
```

#Severity Estimation

```python
def calculate_severity(result, image_width, image_height, class_names,  minor_threshold=5, moderate_threshold=15):

    issues = []

    # Total image area
    image_area = image_width * image_height

    # Check detected damages
    for box in result.boxes:

        # Damage class
        class_id = int(box.cls[0])
        damage_name = class_names[class_id]

        # Bounding box
        x1, y1, x2, y2 = box.xyxy[0].tolist()

        # Damage area
        damage_width = x2 - x1
        damage_height = y2 - y1
        damage_area = damage_width * damage_height

        # Approximate damaged area percentage
        damage_percentage = (damage_area / image_area) * 100

        issues.append({issue = {
    "issue": damage_name,
    "damage_percentage": damage_percentage,
    "accident_description": accident_description
}
        })

    # No damage
    if len(issues) == 0:

        severity = "No Damage"
        total_damage = 0

    else:

        # Total damage percentage
        total_damage = sum(
            item["damage_percentage"]
            for item in issues
        )

        # Severity rules
        if total_damage <  minor_threshold:
            severity = "Minor"

        elif total_damage < moderate_threshold:
            severity = "Moderate"

        else:
            severity = "Severe"

    return {
        "severity": severity,
        "total_damage_percentage": round(total_damage, 2),
        "issues": issues
    }
```

```python
r = results[0]

img_height, img_width = r.orig_shape

severity_result = calculate_severity(
    r,
    img_width,
    img_height,
    data_yaml["names"]
)

print(severity_result)
```

```python
# FUNCTION TO GET DAMAGE DESCRIPTION

def get_damage_description(damage_name, description_index=0):
    """
    Returns an accident/damage description for the detected damage.
    """

    # Convert to lowercase for matching
    damage_name = damage_name.lower().strip()

    if damage_name not in damage_descriptions:
        return "No description available for this damage type."

    descriptions = damage_descriptions[damage_name]

    # Make sure index is valid
    description_index = description_index % len(descriptions)

    return descriptions[description_index]
```
